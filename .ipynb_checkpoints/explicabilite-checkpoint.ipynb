```python
import pandas as pd
import shap
import lightgbm as lgb
import re
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# 1. Chargement et nettoyage des colonnes pour LightGBM
print(" Chargement des données...")
df = pd.read_csv('data/train_cleaned_final.csv').fillna(0)
df.columns = [re.sub(r'[^a-zA-Z0-9_]', '_', col) for col in df.columns]

X = df.drop(columns=['TARGET', 'SK_ID_CURR'])
y = df['TARGET']

# 2. Ton modèle champion issu de l'Étape 4 (Trial 5)
best_params = {
    'objective': 'binary',
    'boosting_type': 'gbdt',
    'class_weight': 'balanced',
    'learning_rate': 0.0463,
    'num_leaves': 84,
    'feature_fraction': 0.677,
    'bagging_fraction': 0.661,
    'min_child_samples': 12,
    'n_estimators': 150,
    'n_jobs': -1,
    'verbosity': -1
}

model = lgb.LGBMClassifier(**best_params)
model.fit(X, y)

# 3. Initialisation de SHAP (on prend 500 clients pour que ce soit rapide)
print(" Calcul des valeurs SHAP...")
explainer = shap.TreeExplainer(model)
X_sample = X.iloc[:500]
shap_values = explainer(X_sample)

# Initialiser l'affichage JavaScript dans le notebook (requis pour SHAP)
shap.initjs()
```

```python
# --- 1. IMPORTANCE GLOBALE ---
print(" IMPORTANCE GLOBALE (Vision Banque)")
shap.plots.bar(shap_values, max_display=10)

# --- 2. SÉLECTION DES CLIENTS ---
probs = model.predict_proba(X_sample)[:, 1]
seuil = 0.51  # Ton seuil optimal trouvé à l'étape 4

# Trouver les index du premier accepté et du premier refusé
idx_accepte = [i for i, p in enumerate(probs) if p < seuil][0]
idx_refuse = [i for i, p in enumerate(probs) if p > seuil][0]

print("\n" + "="*50 + "\n")

# --- 3. IMPORTANCE LOCALE : CLIENT ACCEPTÉ ---
print(f"✅ IMPORTANCE LOCALE : Client ACCEPTÉ (Index {idx_accepte} - Probabilité de risque : {probs[idx_accepte]:.2f})")
shap.plots.waterfall(shap_values[idx_accepte], max_display=10)

print("\n" + "="*50 + "\n")

# --- 4. IMPORTANCE LOCALE : CLIENT REFUSÉ ---
print(f"❌ IMPORTANCE LOCALE : Client REFUSÉ (Index {idx_refuse} - Probabilité de risque : {probs[idx_refuse]:.2f})")
shap.plots.waterfall(shap_values[idx_refuse], max_display=10)
```

```python
# 1. Calcul des coûts pour 81 seuils
seuils = np.linspace(0.1, 0.9, 81)
couts = []

# Extraction des vraies étiquettes pour les 500 premiers clients
y_sample = y.iloc[:500] 

preds_proba = model.predict_proba(X_sample)[:, 1]
for s in seuils:
    preds_seuil = (preds_proba >= s).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_sample, preds_seuil).ravel() if len(np.unique(y_sample)) > 1 else (0,0,0,0)
    
    # Ta fonction de coût métier personnalisée (FN=10, FP=1)
    cout = (fn * 10) + (fp * 1)
    couts.append(cout)

# 2. Tracé de la courbe de coût
plt.figure(figsize=(10, 5))
plt.plot(seuils, couts, color='darkblue', lw=2, marker='o', markevery=5)
plt.axvline(x=0.51, color='red', linestyle='--', label='Seuil Optimal: 0.51')
plt.title("Optimisation du Seuil Décisionnel en fonction du Coût Métier")
plt.xlabel("Seuil de probabilité (Classification Threshold)")
plt.ylabel("Coût Financier Global (Business Cost)")
plt.legend()
plt.grid(True)

# 3. Sauvegarde de l'artifact demandé par la grille d'évaluation
plt.savefig("cost_threshold_curve.png")
plt.show()
print("💾 La courbe cost_threshold_curve.png a été générée et sauvegardée avec succès !")
```
